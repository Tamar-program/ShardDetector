```python
import numpy as np
from PIL import Image
import requests
import cv2
import pytesseract
import matplotlib.pyplot as plt

print("Performing handwriting recognition with Tesseract and OpenCV...")
pytesseract.pytesseract.tesseract_cmd = r'C:\Program Files\Tesseract-OCR\tesseract.exe'
# --- Preparing the image for identification ---
try:
    img_cv = cv2.imread(r'C:\Users\user\Desktop\iaa-pictures\eee.png')
    if img_cv is None:
        raise FileNotFoundError(f"The file was not found or was not loaded correctly..")
    print(f"Loading a local image ")
except FileNotFoundError as e:
    print(f"error: {e}")
    exit() 


# --- Preprocessing with OpenCV ---

# 1. Convert to grayscale image
gray = cv2.cvtColor(img_cv, cv2.COLOR_BGR2GRAY)

# 2. Removing noise using a Gaussian filter
blurred = cv2.GaussianBlur(gray, (5,5), 0)

# 3. Adaptive binarization (critical for handwriting)
binary_img = cv2.adaptiveThreshold(blurred, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
                                   cv2.THRESH_BINARY_INV,9, 6)



# 4.Convert OpenCV image to PIL format (required by pytesseract)
img_pil = Image.fromarray(binary_img)

plt.imshow(img_pil)
plt.axis('off')
plt.show()

# --- Performing the identification with Tesseract---


try:
   
    text = pytesseract.image_to_string(img_pil, lang='eng', config='--psm 13')

    print("\n---Identification results ---")
    print("The recognized text:")
    print(text)

except pytesseract.TesseractNotFoundError:
    print("\nError: Tesseract OCR not found. Make sure you have Tesseract installed as software on your system.")
    print("Or you need to set the path to tesseract.exe using: pytesseract.pytesseract.tesseract_cmd = r'C:\\path\\to\\tesseract.exe'")
except Exception as e:
    print(f"\nAn error occurred while identifying with Tesseract.: {e}")
```

```python
from difflib import SequenceMatcher

# The ground truth
expected = """A9656
14
123"""

# The OCR output
ocr_output = text.strip().replace(" ", "")

# Comparison
matcher = SequenceMatcher(None, expected, ocr_output)
print("Similarity ratio:", matcher.ratio())
```

```python
import numpy as np
from PIL import Image
import requests
import cv2
import pytesseract
import matplotlib.pyplot as plt

print("Performing handwriting recognition with Tesseract and OpenCV...")

# --- Preparing the image for identification ---
try:
    img_cv = cv2.imread(r'C:\Users\user\Desktop\project_folder\images\test\WhatsApp Image 2025-07-13 at 11.01.50_2fe79fc4.jpg')
    if img_cv is None:
        raise FileNotFoundError(f"The file was not found or was not loaded correctly.")
    print(f"Loading a local image ")
except FileNotFoundError as e:
    print(f"error: {e}")
    exit() 


# --- Preprocessing with OpenCV ---

# 1. Convert to grayscale image
gray = cv2.cvtColor(img_cv, cv2.COLOR_BGR2GRAY)

# 2. Removing noise using a Gaussian filter
blurred = cv2.GaussianBlur(gray, (5,5), 0)

# 3. Adaptive binarization (critical for handwriting)
binary_img = cv2.adaptiveThreshold(blurred, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
                                   cv2.THRESH_BINARY_INV,11, 8)



# 4.Convert OpenCV image to PIL format (required by pytesseract)
img_pil = Image.fromarray(binary_img)

plt.imshow(img_pil)
plt.axis('off')
plt.show()

# --- Performing the identification with Tesseract---


# config='--psm 6' או '--psm 7' או '--psm 11' או '--psm 13' - נסי להתנסות
# PSM 6: Assume a single uniform block of text.
# PSM 7: Treat the image as a single text line. (טוב לשורות בודדות)
# PSM 11: Find as much text as possible in no particular order.
# PSM 13: Treat the image as a single text line, bypassing hacks that are Tesseract-specific.
try:
   
    text = pytesseract.image_to_string(img_pil, lang='eng', config='--psm 13')

    print("\n---Identification results ---")
    print("The recognized text:")
    print(text)

except pytesseract.TesseractNotFoundError:
    print("\nError: Tesseract OCR not found. Make sure you have Tesseract installed as software on your system.")
    print("Or you need to set the path to tesseract.exe using: pytesseract.pytesseract.tesseract_cmd = r'C:\\path\\to\\tesseract.exe'")
except Exception as e:
    print(f"\nAn error occurred while identifying with Tesseract.: {e}")
```

```python
# The ground truth
expected = """A9656
14
123"""

# The OCR output
ocr_output = text.strip().replace(" ", "")

# Comparison
matcher = SequenceMatcher(None, expected, ocr_output)
print("Similarity ratio:", matcher.ratio())
```
